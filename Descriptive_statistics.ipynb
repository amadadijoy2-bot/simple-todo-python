Import Statistics

```python
import math

data = [4, 5, 6, 7, 8, 9, 10, 11, 30]
```

Central Tendency

```python
mean_value = sum(data) / len(data)
print("Mean:", round(mean_value, 2))

median_value = sorted(data)[len(data) // 2]
print("Median:", round(median_value, 2))
```

Dispersion

```python
population_variance = sum((x - mean_value) ** 2 for x in data) / len(data)
print("Population Variance:", round(population_variance, 2))

population_standard_deviation = math.sqrt(population_variance)
print("Population Standard Deviation:", round(population_standard_deviation, 2))

sample_variance = sum((x - mean_value) ** 2 for x in data) / (len(data) - 1)
print("Sample Variance:", round(sample_variance, 2))

sample_stdev = math.sqrt(sample_variance)
print("Sample Standard Deviation:", round(sample_stdev, 2))
```

Quartiles using median of halves method

```python
import statistics

Lower_half = data[:len(data) // 2]
Upper_half = data[len(data) // 2 + 1 :]

Q1 = statistics.median(Lower_half)
Q3 = statistics.median(Upper_half)

iqr = Q3 - Q1
lower_fence = Q1 - 1.5 * iqr
upper_fence = Q3 + 1.5 * iqr
outliers = [x for x in data if x < lower_fence or x > upper_fence]


print("Lower Quartile (Q1):", round(Q1, 2))
print("Upper Quartile (Q3):", round(Q3, 2))
print("Interquartile Range (IQR):", round(iqr, 2))
print("Lower Fence:", round(lower_fence, 2))
print("Upper Fence:", round(upper_fence, 2))
print("Outliers:", outliers)
```
