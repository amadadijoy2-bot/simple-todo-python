Import Libraries


```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'County' : ["Nairobi"]*4 + ["Mombasa"]*4 + ["Kisumu"]*4,
    'Month' : ["January", "February", "March", "April"]*3,
    'Latitude' : [-1.29]*4 + [-4.05]*4 + [-0.10]*4,
    'Longitude' : [36.82]*4 + [39.67]*4 + [34.75]*4,
    'Rainfall(mm)' : [35, 48, 90, 140, 25, 40, 110, 180, 60, 75, 130, 160]
}
```

Data Frame

```python
df = pd.DataFrame(data)
print(df.head())
```

Spatial Visualisation

```python
locations = df.drop_duplicates("County")

plt.figure(figsize=(8, 5))
plt.scatter(locations["Longitude"], locations["Latitude"])

for _, row in locations.iterrows():
    plt.annotate(
        row["County"],
        (row["Longitude"], row["Latitude"]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Geographical Locations of Selected Kenyan Counties")
plt.grid(True)
plt.show()
```

Temporal Visualization

```python
months = ["January", "February", "March", "April"]

plt.figure(figsize=(8, 5))
for county in df["County"].unique():
    Subset = df[df["County"] == county].set_index("Month").loc[months]
    plt.plot(months, Subset["Rainfall(mm)"], marker="o", label=county)

plt.xlabel("Month")
plt.ylabel("Rainfall(mm)")
plt.title("Monthly Rainfall Across Selected Kenyan Counties")
plt.legend()
plt.grid(True)
plt.show()
```
