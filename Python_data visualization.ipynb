Create a DataFrame

```python
df = pd.DataFrame({'Height': height, 'Weight': weight})
print (df)
```

Plot Height against Weight

```python
plt.figure(figsize=(8, 5))
plt.plot(df['Weight'], df['Height'], marker='o', linestyle='-',  color='teal' ,
         label='observations')

plt.title('Height against Weight - Region A')
plt.xlabel('Weight')
plt.ylabel('Height')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```
